```python
import pandas as pd
import glob
import os

files = sorted(glob.glob("../data/raw/*.csv"))
frames = []

for f in files:
    season = os.path.basename(f).replace(".csv", "")
    df = pd.read_csv(f, encoding="latin-1")
    df["Season"] = season
    frames.append(df)

matches = pd.concat(frames, ignore_index=True)

cols = ["Season", "Date", "HomeTeam", "AwayTeam",
        "FTHG", "FTAG", "FTR",
        "HS", "AS", "HST", "AST",
        "HC", "AC", "HF", "AF",
        "HY", "AY", "HR", "AR"]
matches = matches[cols]

matches = matches.dropna(subset=["HomeTeam"]).reset_index(drop=True)

print("Shape:", matches.shape)
matches.head()
```
